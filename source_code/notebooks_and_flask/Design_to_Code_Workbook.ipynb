# Part 1 – HTML Basics

**HTML** stands for HyperText Markup Language.

Every webpage you have ever seen is built on HTML. It describes **what** is on the page — headings, paragraphs, buttons, images, links.

HTML does **not** control how things look. That is CSS. HTML is only structure.

### How HTML works

HTML uses **tags** to wrap content:

```
<tagname>  content  </tagname>
    ^                   ^
 opening tag        closing tag
```

Some tags do not need a closing tag:
```html
<img src="photo.jpg">   <!-- image -->
<br>                    <!-- line break -->
<hr>                    <!-- horizontal line -->
```


## 1.1 – The Minimal HTML Page

Every HTML file has this structure. Run the cell below — it creates a file called `page.html` in your folder.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>My First Page</title>
  </head>
  <body>
    <h1>Hello World</h1>
  </body>
</html>
````

**What each line does:**

| Line | Meaning |
|------|---------|
| `<!DOCTYPE html>` | Tells the browser this is an HTML5 page |
| `<html lang="en">` | Root element. `lang="en"` tells the browser the language is English |
| `<head>` | Contains information about the page (not visible to the user) |
| `<title>` | The text shown in the browser tab |
| `<body>` | Everything visible on the page goes here |
| `<h1>` | The largest heading. There should be only one `<h1>` per page |


## 1.2 – Headings and Paragraphs

HTML has six heading sizes: `<h1>` (largest) to `<h6>` (smallest).
Paragraphs use the `<p>` tag.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head><title>Headings</title></head>
  <body>

    <h1>This is H1 – Page Title</h1>
    <h2>This is H2 – Section Title</h2>
    <h3>This is H3 – Sub-section</h3>

    <p>This is a paragraph. It holds regular body text.</p>
    <p>This is a second paragraph. Each &lt;p&gt; tag starts on a new line.</p>

  </body>
</html>
````

## 1.3 – Links and Images

Links use the `<a>` tag. The destination URL goes in the `href` attribute.
Images use the `<img>` tag. The file path or URL goes in the `src` attribute.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head><title>Links and Images</title></head>
  <body>

    <!-- Link that opens in a new tab -->
    <a href="https://github.com" target="_blank">Visit GitHub</a>

    <!-- Image from the internet -->
    <img
      src="https://via.placeholder.com/120"
      alt="A placeholder image"
      width="120"
    />

  </body>
</html>
````

## 1.4 – Divs, Classes, and IDs

A `<div>` is a generic container — a box that holds other elements.
You label elements using `class` (many elements can share a class) or `id` (unique, only one per page).
CSS and JavaScript use these labels to target specific elements.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head><title>Divs and Classes</title></head>
  <body>

    <!-- A named box -->
    <div class="card">
      <h2 id="user-name">Priya Sharma</h2>
      <p class="bio">2nd year B.Tech CSE student.</p>
    </div>

    <!-- Another box with the same class -->
    <div class="card">
      <h2>Rohit Desai</h2>
      <p class="bio">Passionate about machine learning.</p>
    </div>

  </body>
</html>
````

| Attribute | Meaning |
|-----------|--------|
| `class="card"` | Applies the name `card` to this element. Multiple elements can share the same class. CSS uses `.card` to style all of them at once. |
| `id="user-name"` | A unique identifier. Only one element on the page should have this id. CSS uses `#user-name`. |


## 1.5 – Buttons and Inputs

Buttons and input fields make pages interactive.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head><title>Buttons and Inputs</title></head>
  <body>

    <!-- Text input -->
    <label for="username">Your name:</label>
    <input type="text" id="username" placeholder="Enter your name" />

    <!-- Number input -->
    <label for="age">Your age:</label>
    <input type="number" id="age" placeholder="e.g. 20" />

    <!-- Button -->
    <button>Submit</button>

  </body>
</html>
````

# Part 2 – CSS Basics

**CSS** stands for Cascading Style Sheets.

If HTML is the skeleton of a page, CSS is the skin — colors, fonts, spacing, layout.

### CSS Syntax

```css
selector {
  property: value;
}
```

The **selector** targets one or more HTML elements.
The **property** is what you want to change.
The **value** is what you want it to become.

### Three ways to write CSS

| Method | How | Use case |
|--------|-----|----------|
| Inline | `<h1 style="color: red;">` | Quick test only |
| Internal | `<style>` tag inside `<head>` | Small single pages |
| External | Separate `style.css` file, linked with `<link>` | **Always use this in real projects** |


## 2.1 – Colors and Fonts

The most common starting point: changing text color, background color, and font.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>Colors and Fonts</title>
    <style>

      body {
        background-color: #f0f4ff;   /* light blue background */
        font-family: Arial, sans-serif;
      }

      h1 {
        color: #4f46e5;              /* indigo */
        font-size: 2rem;
      }

      p {
        color: #374151;              /* dark gray */
        font-size: 1rem;
      }

    </style>
  </head>
  <body>
    <h1>Hello CSS</h1>
    <p>This paragraph is styled with CSS.</p>
  </body>
</html>
````

**Key properties used:**

| Property | What it controls |
|----------|------------------|
| `background-color` | The fill color behind the element |
| `color` | The text color |
| `font-family` | The typeface. Always add a fallback: `Arial, sans-serif` |
| `font-size` | Size of the text. `rem` = relative to the root font size (16px by default) |

**Tip:** CSS colors can be written as:
- Hex codes: `#4f46e5`
- RGB: `rgb(79, 70, 229)`
- Named: `red`, `blue`, `coral`


## 2.2 – Spacing: Padding and Margin

These are the two most important spacing properties in CSS.

```
+------------------------------------------+
|              MARGIN (outside)             |
|   +----------------------------------+    |
|   |          PADDING (inside)        |    |
|   |   +-------------------------+   |    |
|   |   |   CONTENT (text/image)  |   |    |
|   |   +-------------------------+   |    |
|   +----------------------------------+    |
+------------------------------------------+
```

- **Padding** = space INSIDE the element (between the border and the content)
- **Margin** = space OUTSIDE the element (between this element and others)


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>Padding and Margin</title>
    <style>

      .box {
        background-color: #c7d2fe;
        padding: 24px;            /* space inside the box */
        margin: 32px;             /* space outside the box */
        border: 2px solid #4f46e5;
        border-radius: 8px;       /* rounded corners */
      }

    </style>
  </head>
  <body>
    <div class="box">Box A – I have padding and margin</div>
    <div class="box">Box B – Same class, same style</div>
  </body>
</html>
````

## 2.3 – Layout with Flexbox

By default, HTML elements stack vertically. **Flexbox** lets you arrange them horizontally, center them, and control spacing — all with a few CSS lines.

You apply `display: flex` to the **parent** element. The children are arranged automatically.


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>Flexbox</title>
    <style>

      body {
        /* Center everything on the page */
        min-height: 100vh;
        display: flex;
        align-items: center;
        justify-content: center;
        background-color: #f0f4ff;
        margin: 0;
      }

      .row {
        display: flex;            /* children go side by side */
        gap: 16px;                /* space between children */
      }

      .pill {
        background-color: #4f46e5;
        color: white;
        padding: 10px 20px;
        border-radius: 999px;     /* fully rounded */
        font-family: Arial, sans-serif;
      }

    </style>
  </head>
  <body>
    <div class="row">
      <div class="pill">HTML</div>
      <div class="pill">CSS</div>
      <div class="pill">Figma</div>
    </div>
  </body>
</html>
````

**Flexbox quick reference:**

| Property | Values | Effect |
|----------|--------|--------|
| `display: flex` | — | Turns the element into a flex container |
| `flex-direction` | `row` (default) / `column` | Horizontal or vertical |
| `justify-content` | `center` / `space-between` / `flex-start` | Main axis alignment |
| `align-items` | `center` / `flex-start` / `stretch` | Cross axis alignment |
| `gap` | `8px`, `16px`… | Space between children |


## 2.4 – External CSS File

In real projects, CSS lives in a separate `.css` file. You link it inside the `<head>` tag.
This keeps HTML clean (structure) and CSS separate (appearance).


````python
%%writefile page.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>External CSS</title>
    <link rel="stylesheet" href="style.css" />
  </head>
  <body>
    <div class="card">
      <h1 class="name">Priya Sharma</h1>
      <p class="role">B.Tech CSE Student</p>
    </div>
  </body>
</html>
````

````python
%%writefile style.css
body {
  background-color: #eef2ff;
  display: flex;
  justify-content: center;
  align-items: center;
  min-height: 100vh;
  margin: 0;
  font-family: Arial, sans-serif;
}

.card {
  background-color: white;
  padding: 40px;
  border-radius: 16px;
  text-align: center;
  box-shadow: 0 4px 20px rgba(0,0,0,0.1);
}

.name {
  color: #1e1b4b;
  font-size: 1.8rem;
}

.role {
  color: #6b7280;
  font-size: 1rem;
}
````

> Open both files in VS Code. Double-click `page.html` to preview in the browser. Change the `background-color` in `style.css`, save, and refresh — the page updates instantly.

**Remember:** The `href` in `<link href="style.css">` must exactly match the filename. `style.css` and `Style.css` are treated as different files on Vercel (Linux servers are case-sensitive).

# Part 3 – Figma: Theory


## 3.1 – What is Figma?

**Figma** is a design tool that runs in the browser. It is used to design the visual appearance of apps and websites **before writing any code**.

Figma is free for individual use. You do not need to install anything — it works in Chrome or Edge.

Real-world context:
- Every major app you use — Instagram, Swiggy, Google Pay, Zepto — was designed in Figma (or a similar tool) before a single line of code was written.
- Designers use Figma. Developers use the designs as a blueprint.
- Many developers also use Figma to plan their own projects.


## 3.2 – Why Use Figma?

| Without Figma | With Figma |
|---------------|------------|
| You write code and guess how it looks | You see exactly what you are building |
| Changing a layout requires rewriting CSS | Drag and drop in Figma |
| Hard to plan spacing, fonts, and colors | Everything is measured and consistent |
| No way to show a client before building | Share a link — client clicks through the design |

**The rule:** Design first, then code. This saves hours of rework.


## 3.3 – How to Use Figma

### Step 1 – Open Figma
Go to [figma.com](https://figma.com), sign in, click **New Design File**.

### Step 2 – The interface

```
+------------------------------------------------------+
|  TOP BAR  -- tools, file name, share button, zoom   |
+-----------+---------------------------+--------------+
|           |                           |              |
| LEFT      |       CANVAS              | RIGHT        |
| PANEL     |  (your design lives here) | PANEL        |
|           |                           |              |
| Layers    |                           | Properties   |
| Pages     |                           | Colors       |
| Assets    |                           | Spacing      |
+-----------+---------------------------+--------------+
```

### Step 3 – Basic tools

| Tool | Shortcut | What it creates |
|------|----------|-----------------|
| Frame | `F` | A screen or section container |
| Rectangle | `R` | A box (for cards, buttons, avatars) |
| Ellipse | `O` | A circle (for avatars, icons) |
| Text | `T` | A text element |
| Move/Select | `V` | Selects and moves elements |

### Step 4 – Right panel (Properties)

When you select any element, the right panel shows:
- **W / H** — width and height in pixels
- **X / Y** — position on the canvas
- **Corner radius** — how rounded the corners are
- **Fill** — background color
- **Stroke** — border color
- **Effects** — shadows

### Step 5 – Auto Layout

Auto Layout is Figma's version of CSS Flexbox.

Select an element → press `Shift+A` → the element becomes a flex container.
You can then set: direction (row/column), gap between children, and padding.

### Step 6 – Inspect (the bridge to code)

Select any element → click the **Inspect** tab in the right panel.
Figma shows you the CSS values for that element:

```css
/* Example: Figma Inspect output for a button */
width: 180px;
height: 48px;
background: #4F46E5;
border-radius: 8px;
font-size: 16px;
font-weight: 600;
color: #FFFFFF;
```

You copy these values directly into your CSS file.


## 3.4 – Figma Concepts Mapped to HTML + CSS

| Figma | HTML | CSS |
|-------|------|-----|
| Frame | `<div>` or `<section>` | `display: flex` |
| Rectangle | `<div>` | `background-color`, `border-radius` |
| Ellipse (circle) | `<div>` | `border-radius: 50%` |
| Text | `<h1>`, `<p>`, `<span>` | `font-size`, `font-weight`, `color` |
| Button (rectangle + text) | `<button>` or `<a>` | `padding`, `border-radius`, `background-color` |
| Fill color | — | `background-color` |
| Text color | — | `color` |
| Corner radius | — | `border-radius` |
| Stroke | — | `border: Npx solid #hex` |
| Drop shadow | — | `box-shadow: 0 4px 16px rgba(0,0,0,0.1)` |
| Auto Layout (vertical) | — | `display: flex; flex-direction: column; gap: Npx` |
| Auto Layout (horizontal) | — | `display: flex; flex-direction: row; gap: Npx` |
| Padding (Auto Layout) | — | `padding: Npx` |
| Gap (Auto Layout) | — | `gap: Npx` |

# Part 4 – From Figma Design to Real Code

In this section we take a Figma design of a **Profile Card** and convert it step by step into HTML + CSS.

Imagine you designed this card in Figma:

```
+-----------------------------+
|                             |
|          ( P S )            |  <- circle, 90x90, fill #4F46E5
|                             |
|       Priya Sharma          |  <- text, 24px, bold, #1e1b4b
|   B.Tech CSE  PICT Pune     |  <- text, 14px, #6b7280
|  ------- divider --------   |  <- stroke, #e5e7eb
|  Passionate about the web.  |  <- text, 14px, #374151
|                             |
|  [Python] [CSS] [Figma]     |  <- tags, fill #ede9fe, text #4f46e5
|                             |
|  [GitHub]     [LinkedIn]    |  <- buttons, fill #4f46e5, text white
|                             |
+-----------------------------+
   card: 400x520, fill white
   corner radius: 20
   drop shadow: 0 8 40 rgba(79,70,229,0.13)
   page background: #eef2ff
```

Now let us translate each Figma element into code.


## 4.1 – HTML Structure (from Figma layout)

Each Figma layer becomes an HTML element. We follow the **same nesting order** as the Figma layer panel.


````python
%%writefile profile-card/index.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <meta charset="UTF-8" />
    <meta name="viewport" content="width=device-width, initial-scale=1.0" />
    <title>Profile Card</title>
    <link rel="stylesheet" href="style.css" />
    <!-- Google Font: Inter (same font used in Figma design) -->
    <link
      href="https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&display=swap"
      rel="stylesheet"
    />
  </head>
  <body>

    <!-- PAGE: Figma frame (1440px desktop, centered) -->
    <main class="page">

      <!-- CARD: Figma frame (400px, white, r=20, shadow) -->
      <div class="card">

        <!-- AVATAR: Figma ellipse (90x90, fill #4f46e5) -->
        <div class="avatar">PS</div>

        <!-- NAME: Figma text (24px, 700, #1e1b4b) -->
        <h1 class="name">Priya Sharma</h1>

        <!-- ROLE: Figma text (14px, 400, #6b7280) -->
        <p class="role">B.Tech CSE &bull; PICT College, Pune</p>

        <!-- DIVIDER: Figma stroke (1px, #e5e7eb) -->
        <hr class="divider" />

        <!-- BIO: Figma text (14px, 400, #374151) -->
        <p class="bio">
          Passionate about building things for the web.
          Currently in 2nd year and learning every day.
        </p>

        <!-- SKILLS: Figma Auto Layout row, gap 8 -->
        <div class="skills">
          <!-- TAGS: Figma rectangle (fill #ede9fe, r=999, text #4f46e5) -->
          <span class="tag">Python</span>
          <span class="tag">HTML</span>
          <span class="tag">CSS</span>
          <span class="tag">Figma</span>
          <span class="tag">Git</span>
        </div>

        <!-- LINKS: Figma Auto Layout row, gap 12 -->
        <div class="links">
          <!-- BUTTON: Figma rectangle (fill #4f46e5, r=9, text white) -->
          <a class="btn-primary" href="https://github.com/" target="_blank">GitHub</a>
          <a class="btn-secondary" href="https://linkedin.com/" target="_blank">LinkedIn</a>
        </div>

      </div>
    </main>

  </body>
</html>
````

**How each Figma layer became an HTML element:**

| Figma Layer | HTML Tag | Class name | Reason |
|-------------|----------|------------|--------|
| Page frame | `<main>` | `page` | Outer container |
| Card frame | `<div>` | `card` | The white box |
| Avatar ellipse | `<div>` | `avatar` | Will become a circle in CSS |
| Name text | `<h1>` | `name` | Heading (most important text) |
| Role text | `<p>` | `role` | Paragraph |
| Stroke line | `<hr>` | `divider` | Horizontal rule |
| Bio text | `<p>` | `bio` | Paragraph |
| Skills frame | `<div>` | `skills` | Row of tags |
| Skill tag rectangles | `<span>` | `tag` | Inline boxes |
| Links frame | `<div>` | `links` | Row of buttons |
| Button rectangles | `<a>` | `btn-primary/secondary` | Clickable links styled as buttons |


## 4.2 – CSS (direct values from Figma Inspect)

Every value below came straight from the Figma Inspect panel.
Comments show the exact Figma property → CSS property mapping.


````python
%%writefile profile-card/style.css
/* ================================================
   CSS VARIABLES  (Figma: Design tokens / styles)
   Change these to retheme the card instantly.
================================================ */
:root {
  --accent:       #4f46e5;    /* Figma: primary color style */
  --accent-dark:  #4338ca;
  --accent-soft:  #ede9fe;    /* Figma: secondary color style */
  --bg:           #eef2ff;    /* Figma: page background fill */
  --card-bg:      #ffffff;    /* Figma: card fill */
  --text-main:    #1e1b4b;    /* Figma: heading text color */
  --text-body:    #374151;    /* Figma: body text color */
  --text-muted:   #6b7280;    /* Figma: muted text color */
  --border:       #e5e7eb;    /* Figma: divider stroke */
}

/* ================================================
   RESET
================================================ */
*, *::before, *::after {
  margin: 0; padding: 0; box-sizing: border-box;
}

/* ================================================
   PAGE  (Figma: desktop frame, centers the card)
================================================ */
body { font-family: 'Inter', Arial, sans-serif; background-color: var(--bg); }

.page {
  min-height: 100vh;
  display: flex;               /* Figma: Auto Layout */
  align-items: center;         /* Figma: align = center */
  justify-content: center;     /* Figma: justify = center */
  padding: 24px;
}

/* ================================================
   CARD  (Figma: frame 400w, fill #fff, r=20, shadow)
================================================ */
.card {
  background-color: var(--card-bg); /* Figma: fill */
  width: 100%;
  max-width: 400px;            /* Figma: width = 400 */
  border-radius: 20px;         /* Figma: corner radius = 20 */
  padding: 44px 40px;          /* Figma: padding = 44 top/bottom, 40 left/right */
  /* Figma: drop shadow = 0 8 40 rgba(79,70,229,0.13) */
  box-shadow: 0 8px 40px rgba(79, 70, 229, 0.13);
  display: flex;               /* Figma: Auto Layout */
  flex-direction: column;      /* Figma: direction = vertical */
  align-items: center;         /* Figma: align = center */
  gap: 18px;                   /* Figma: gap = 18 */
  text-align: center;
}

/* ================================================
   AVATAR  (Figma: ellipse 90x90, fill #4f46e5)
================================================ */
.avatar {
  width: 90px;                 /* Figma: W = 90 */
  height: 90px;                /* Figma: H = 90 */
  border-radius: 50%;          /* Makes the rectangle a circle */
  background-color: var(--accent); /* Figma: fill = #4f46e5 */
  color: #fff;
  font-size: 1.4rem;
  font-weight: 700;
  display: flex;
  align-items: center;
  justify-content: center;
}

/* ================================================
   NAME  (Figma: text, 24px, 700, #1e1b4b)
================================================ */
.name {
  font-size: 1.5rem;           /* Figma: font size = 24px */
  font-weight: 800;            /* Figma: font weight = Bold */
  color: var(--text-main);     /* Figma: fill = #1e1b4b */
}

/* ================================================
   ROLE  (Figma: text, 14px, 400, #6b7280)
================================================ */
.role {
  font-size: 0.875rem;         /* Figma: font size = 14px */
  color: var(--text-muted);    /* Figma: fill = #6b7280 */
  margin-top: -8px;
}

/* ================================================
   DIVIDER  (Figma: stroke 1px, #e5e7eb)
================================================ */
.divider {
  width: 100%;
  border: none;
  border-top: 1px solid var(--border); /* Figma: stroke */
}

/* ================================================
   BIO  (Figma: text, 14px, 400, #374151)
================================================ */
.bio {
  font-size: 0.9rem;           /* Figma: font size = 14px */
  color: var(--text-body);     /* Figma: fill = #374151 */
  line-height: 1.65;
  max-width: 300px;
}

/* ================================================
   SKILLS  (Figma: Auto Layout, row, gap 8)
================================================ */
.skills {
  display: flex;               /* Figma: Auto Layout */
  flex-wrap: wrap;
  justify-content: center;
  gap: 8px;                    /* Figma: gap = 8 */
}

/* TAG  (Figma: rectangle, fill #ede9fe, r=999, text #4f46e5, 12px, 600) */
.tag {
  background-color: var(--accent-soft); /* Figma: fill = #ede9fe */
  color: var(--accent);                 /* Figma: text fill = #4f46e5 */
  font-size: 0.78rem;                   /* Figma: font size = 12px */
  font-weight: 600;                     /* Figma: font weight = SemiBold */
  padding: 5px 14px;                    /* Figma: padding = 5 top/bottom, 14 left/right */
  border-radius: 999px;                 /* Figma: corner radius = 999 (fully rounded) */
}

/* ================================================
   LINKS  (Figma: Auto Layout, row, gap 12)
================================================ */
.links {
  display: flex;               /* Figma: Auto Layout */
  gap: 12px;                   /* Figma: gap = 12 */
  flex-wrap: wrap;
  justify-content: center;
  margin-top: 4px;
}

/* BUTTON  (Figma: rectangle, fill #4f46e5, r=9, text white, 14px, 600) */
.btn-primary, .btn-secondary {
  text-decoration: none;
  padding: 10px 26px;          /* Figma: padding = 10 top/bottom, 26 left/right */
  border-radius: 9px;          /* Figma: corner radius = 9 */
  font-size: 0.875rem;         /* Figma: font size = 14px */
  font-weight: 600;            /* Figma: font weight = SemiBold */
  transition: background-color 0.18s;
}

.btn-primary {
  background-color: var(--accent); /* Figma: fill = #4f46e5 */
  color: #ffffff;                  /* Figma: text fill = white */
}
.btn-primary:hover { background-color: var(--accent-dark); }

.btn-secondary {
  background-color: var(--accent-soft); /* Figma: fill = #ede9fe */
  color: var(--accent);                 /* Figma: text fill = #4f46e5 */
}
.btn-secondary:hover { background-color: #ddd6fe; }

/* ================================================
   RESPONSIVE  (not in Figma, added for real-world use)
================================================ */
@media (max-width: 480px) {
  .card     { padding: 32px 20px; }
  .name     { font-size: 1.3rem; }
  .btn-primary, .btn-secondary { padding: 10px 18px; }
}
````

**Key takeaways from the Figma → CSS translation:**

1. **Figma Fill → CSS `background-color` or `color`** — Fill on a shape = `background-color`. Fill on text = `color`.
2. **Figma Corner Radius → CSS `border-radius`** — Exact same number. `20` in Figma → `border-radius: 20px`.
3. **Figma Auto Layout → CSS Flexbox** — Direction, gap, and padding values transfer directly.
4. **Figma Drop Shadow → CSS `box-shadow`** — The values (x, y, blur, spread, color) map 1:1.
5. **Figma Font Size → CSS `font-size`** — Same number, add `px`. Alternatively convert to `rem` (divide by 16).
6. **Figma does not write the responsive CSS for you** — You add `@media` queries yourself.


## 4.3 – Adding More Details: Hover Effects and Transitions

Figma shows static designs. But in real code, buttons react when you hover over them.
CSS `transition` makes the change smooth instead of instant.


````python
%%writefile hover_demo.html
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>Hover Effects</title>
    <style>

      body {
        display: flex; gap: 16px;
        justify-content: center; align-items: center;
        min-height: 100vh; background: #eef2ff; margin: 0;
      }

      /* Button WITHOUT transition -- changes instantly */
      .btn-a {
        padding: 12px 28px;
        background-color: #4f46e5;
        color: white;
        border: none;
        border-radius: 8px;
        font-size: 1rem;
        cursor: pointer;
      }
      .btn-a:hover { background-color: #818cf8; }

      /* Button WITH transition -- changes smoothly */
      .btn-b {
        padding: 12px 28px;
        background-color: #4f46e5;
        color: white;
        border: none;
        border-radius: 8px;
        font-size: 1rem;
        cursor: pointer;
        transition: background-color 0.25s, transform 0.1s;
      }
      .btn-b:hover   { background-color: #818cf8; }
      .btn-b:active  { transform: scale(0.96); }

    </style>
  </head>
  <body>
    <button class="btn-a">No Transition</button>
    <button class="btn-b">With Transition</button>
  </body>
</html>
````

Open `hover_demo.html` in your browser. Hover over both buttons. Notice the difference.

- `transition: background-color 0.25s` — the color changes over 0.25 seconds instead of instantly.
- `transition: transform 0.1s` — the shrink on click (`:active`) is also animated.
- You can transition any CSS property: `color`, `opacity`, `width`, `border-radius`, etc.

## Error Game 1 – HTML

**Find the bug.** The page should show a heading and a paragraph.


````python
# This is the broken HTML -- find the error!
broken_html = """
<!DOCTYPE html>
<html>
  <head><title>Game 1</title></head>
  <body>
    <h1>Welcome to the Bootcamp
    <p>This is the first paragraph.</p>
  </body>
</html>
"""
print(broken_html)
````

<details>

**Bug:** The `<h1>` tag is never closed.

```html
<!-- Broken -->

<!-- Fixed -->
```

**Why it matters:** Without the closing `</h1>`, the browser may treat the `<p>` tag as part of the heading, or render the page incorrectly. Always close your tags.

## Error Game 2 – CSS Selector

**Find the bug.** The heading should turn indigo, but it stays black.


````python
# Find the CSS error!
broken_css = """
/* HTML: <h1 class="title">Hello</h1> */

.Title {
  color: #4f46e5;
  font-size: 2rem;
}
"""
print(broken_css)
````

<details>
<summary><strong>Click to reveal the answer</strong></summary>

**Bug:** CSS class names are case-sensitive. The class is `title` but the selector is `.Title`.

```css
/* Broken */
.Title { color: #4f46e5; }

/* Fixed */
.title { color: #4f46e5; }
```

**Rule:** Always match the case exactly. If the HTML says `class="title"`, write `.title` in CSS, not `.Title` or `.TITLE`.
</details>


## Error Game 3 – CSS Link

**Find the bug.** The CSS file exists but the styles are not applying.


````python
# Find the bug in the <head> section!
broken_html = """
<!DOCTYPE html>
<html lang="en">
  <head>
    <title>My Card</title>
    <link rel="stylesheet" href="styles.css" />
  </head>
  <body>
    <div class="card">Hello</div>
  </body>
</html>

<!-- The CSS file on disk is named: style.css -->
"""
print(broken_html)
````

<details>
<summary><strong>Click to reveal the answer</strong></summary>

**Bug:** The `href` says `styles.css` (with an `s`) but the actual file is `style.css` (no `s`).

```html
<!-- Broken -->
<link rel="stylesheet" href="styles.css" />

<!-- Fixed -->
<link rel="stylesheet" href="style.css" />
```

**Why it matters:** The filename must match exactly — including capitalisation and spelling. This is one of the most common errors beginners make.
</details>


## Error Game 4 – Flexbox

**Find the bug.** The three boxes should sit side by side, but they are still stacking vertically.


````python
# Find the CSS error!
broken_css = """
/* HTML:
<div class="row">
  <div class="box">A</div>
  <div class="box">B</div>
  <div class="box">C</div>
</div>
*/

.box {
  display: flex;
  gap: 16px;
}
"""
print(broken_css)
````

<details>
<summary><strong>Click to reveal the answer</strong></summary>

**Bug:** `display: flex` was applied to `.box` (the children) instead of `.row` (the parent).

```css
/* Broken */
.box { display: flex; gap: 16px; }

/* Fixed */
.row { display: flex; gap: 16px; }
```

**Rule:** `display: flex` is always applied to the **parent** container. The children (`.box`) arrange themselves automatically.
</details>


## Error Game 5 – CSS Value

**Find the bug.** The card should have a white background, but it is transparent.


````python
# Find the CSS error!
broken_css = """
.card {
  background-color: white
  border-radius: 16px;
  padding: 40px;
}
"""
print(broken_css)
````

<details>
<summary><strong>Click to reveal the answer</strong></summary>

**Bug:** The semicolon is missing after `background-color: white`.

**Why it matters:** Without the semicolon, the browser merges the two lines and cannot parse either property correctly. Every CSS declaration must end with a semicolon.
</details>
