## Section 4 – Figma Fundamentals
### 10:10 – 10:55 (45 minutes)

Figma is a browser-based design tool used by designers and developers worldwide.

The key idea: **design the layout before writing any code**. Changes in Figma are fast and free. Changes in code take much longer.

### Opening Figma

1. Go to [figma.com](https://figma.com) and sign in.
2. Click **New Design File**.

### The Figma Interface

```
+----------------------------------------------+
|  TOP BAR  -- Tools, file name, share, zoom  |
+----------+-------------------+---------------+
|          |                   |               |
|  LEFT    |     CANVAS        |    RIGHT      |
|  PANEL   |  (design lives    |    PANEL      |
|  Layers  |   here)           |  Properties   |
|  Pages   |                   |  Colors       |
|  Assets  |                   |  Fonts        |
+----------+-------------------+---------------+
```

### Core Concepts

**Frames** — Press `F`, then click and drag on the canvas to create a frame. A frame represents a screen. Use preset sizes in the right panel (Desktop: 1440 x 900, Mobile: 375 x 812).

**Shapes**
- Rectangle: press `R`
- Ellipse / circle: press `O`

**Text** — Press `T`, click on canvas, type. Use the right panel to change font, size, weight, color, and alignment.

**Layers** — Every element is a layer. Double-click a layer name in the left panel to rename it. Keep layers organized.

**Colors** — Select element → right panel → Fill → click the color swatch → enter a hex code (e.g. `#4F46E5`).

### Auto Layout

Auto Layout automatically resizes and reflows elements — exactly like CSS Flexbox.

To add: select a frame or group → press `Shift+A` → set Direction, Gap, Padding.

Without Auto Layout:
```
Increase padding  ->  text overflows the button
```
With Auto Layout:
```
Increase padding  ->  button grows automatically
```

### Figma to CSS Mapping

| Figma | CSS | Notes |
|-------|-----|-------|
| Auto Layout (vertical) | `display: flex; flex-direction: column` | Stacks children vertically |
| Auto Layout (horizontal) | `display: flex; flex-direction: row` | Places children side by side |
| Gap | `gap: 16px` | Space between children |
| Padding | `padding: 16px` | Space inside the container |
| Corner radius | `border-radius: 12px` | Rounded corners |
| Fill color | `background-color: #hex` or `color: #hex` | Background or text color |
| Width / Height | `width: 400px` / `height: 200px` | Element size |
| Stroke | `border: 1px solid #hex` | Outline around element |
| Drop shadow | `box-shadow: 0 4px 16px rgba(0,0,0,0.1)` | Shadow beneath element |

### Useful Shortcuts

| Action | Shortcut |
|--------|----------|
| Frame tool | `F` |
| Rectangle | `R` |
| Ellipse | `O` |
| Text | `T` |
| Auto Layout | `Shift+A` |
| Copy properties | `Ctrl+Alt+C` |
| Paste properties | `Ctrl+Alt+V` |

### Code 2 – `index.html`

This is the structure of your page. Every element you designed in Figma becomes an HTML element here.

Update the placeholder text with your own name, role, bio, skills, and links before pushing to GitHub.

### Code 3 – `style.css`

All visual design lives here. The colors, fonts, spacing, and layout are controlled by **CSS custom properties** (variables) declared at the top — edit just those values to change the entire theme in seconds.

| Variable | Controls |
|----------|----------|
| `--accent` | Avatar background, button color, tag color |
| `--bg` | Page background color |
| `--card` | Card background color |
| `--text` | Main text color |
| `--muted` | Secondary text color |
| `--radius` | Corner rounding for the card |


````python
%%writefile profile-card/style.css
/* =====================================================
   CSS CUSTOM PROPERTIES (variables)
   Change these values to instantly retheme the card.
===================================================== */
:root {
  --accent:      #4f46e5;          /* indigo           */
  --accent-dark: #4338ca;
  --accent-soft: #ede9fe;
  --bg:          #eef2ff;          /* page background  */
  --card:        #ffffff;          /* card background  */
  --text:        #111827;          /* main text        */
  --muted:       #6b7280;          /* secondary text   */
  --border:      #e5e7eb;
  --shadow:      0 8px 40px rgba(79, 70, 229, 0.13);
  --radius:      20px;
  --font:        'Inter', sans-serif;
}

/* =====================================================
   RESET
===================================================== */
*, *::before, *::after {
  margin: 0;
  padding: 0;
  box-sizing: border-box;
}

/* =====================================================
   PAGE LAYOUT
   Centers the card vertically and horizontally.
===================================================== */
body {
  font-family: var(--font);
  background-color: var(--bg);
  color: var(--text);
  min-height: 100vh;
}

.page {
  min-height: 100vh;
  display: flex;
  align-items: center;
  justify-content: center;
  padding: 24px;
}

/* =====================================================
   CARD
   The white box — all content lives inside.
===================================================== */
.card {
  background-color: var(--card);
  border-radius: var(--radius);
  padding: 44px 40px;
  width: 100%;
  max-width: 420px;
  box-shadow: var(--shadow);

  /* Flexbox: stack children vertically, centered */
  display: flex;
  flex-direction: column;
  align-items: center;
  gap: 18px;
  text-align: center;
}

/* =====================================================
   AVATAR
   Circle showing the user's initials.
   To use a real photo, see the comment below.
===================================================== */
.avatar {
  width: 96px;
  height: 96px;
  border-radius: 50%;
  background-color: var(--accent);
  color: #fff;
  font-size: 1.6rem;
  font-weight: 700;
  display: flex;
  align-items: center;
  justify-content: center;
  letter-spacing: 0.05em;
  user-select: none;
}

/* To use a real photo:
   1. Replace <div class="avatar">PS</div> in HTML with
      <img src="photo.jpg" alt="Your name" class="avatar-img">
   2. Add this rule: */
/* .avatar-img { width:96px; height:96px; border-radius:50%; object-fit:cover; } */

/* =====================================================
   NAME & ROLE
===================================================== */
.name {
  font-size: 1.55rem;
  font-weight: 800;
  color: var(--text);
  line-height: 1.2;
  margin-top: 4px;
}

.role {
  font-size: 0.875rem;
  color: var(--muted);
  margin-top: -10px;
}

/* =====================================================
   DIVIDER
===================================================== */
.divider {
  width: 100%;
  border: none;
  border-top: 1px solid var(--border);
}

/* =====================================================
   BIO
===================================================== */
.bio {
  font-size: 0.9rem;
  color: #374151;
  line-height: 1.65;
  max-width: 320px;
}

/* =====================================================
   SKILLS  —  row of pill-shaped tags
===================================================== */
.skills {
  display: flex;
  flex-wrap: wrap;
  justify-content: center;
  gap: 8px;
}

.tag {
  background-color: var(--accent-soft);
  color: var(--accent);
  font-size: 0.78rem;
  font-weight: 600;
  padding: 5px 14px;
  border-radius: 999px;
  letter-spacing: 0.02em;
}

/* =====================================================
   LINKS  —  two side-by-side buttons
===================================================== */
.links {
  display: flex;
  gap: 12px;
  flex-wrap: wrap;
  justify-content: center;
  margin-top: 4px;
}

.link-btn {
  display: inline-block;
  text-decoration: none;
  padding: 10px 26px;
  border-radius: 9px;
  font-size: 0.875rem;
  font-weight: 600;
  font-family: var(--font);
  transition: background-color 0.18s, transform 0.1s;
}

.link-btn.primary {
  background-color: var(--accent);
  color: #ffffff;
}
.link-btn.primary:hover  { background-color: var(--accent-dark); }

.link-btn.secondary {
  background-color: var(--accent-soft);
  color: var(--accent);
}
.link-btn.secondary:hover { background-color: #ddd6fe; }

.link-btn:active { transform: scale(0.97); }

/* =====================================================
   RESPONSIVE  —  small screens
===================================================== */
@media (max-width: 480px) {
  .card       { padding: 32px 24px; }
  .name       { font-size: 1.3rem; }
  .link-btn   { padding: 10px 20px; }
}
````

## Section 11 – Troubleshooting

### Before Asking for Help

Work through these questions first:

1. What did I expect to happen?
2. What actually happened?
3. Is there an error message?
4. What was the last thing I changed?
5. Can I undo that last change and test again?

### Common Problems

---

**CSS not applying / page looks unstyled**

- Is `<link rel="stylesheet" href="style.css">` inside the `<head>` tag?
- Is the filename spelled exactly right? `style.css` is not the same as `Style.css`.
- Are `index.html` and `style.css` in the same folder?
- Did you save the file after making changes? (`Ctrl+S`)

---

**Changes not showing in the browser**

- Press `Ctrl+S` to save.
- Press `Ctrl+Shift+R` to hard-refresh (bypasses cache).
- If using Live Server, it should auto-refresh.

---

**Photo not showing**

- Is the image file in the same folder as `index.html`?
- Is the filename in `src="photo.jpg"` spelled exactly the same as the actual file name?
- Is the file extension correct (`.jpg`, `.png`, `.webp`)?

---

**`git` is not recognized as a command**

1. Download Git from [git-scm.com](https://git-scm.com) and install (accept all defaults).
2. Restart your terminal.
3. Run `git --version` again.

---

**`git push` fails / asks for a password**

GitHub no longer accepts plain passwords for HTTPS pushes.

Fix: create a Personal Access Token.
1. GitHub → Settings → Developer settings → Personal Access Tokens → Classic.
2. Generate a new token with scope: `repo`.
3. Use the token as the password when Git prompts.

Ask the second instructor for help with this step.

---

**Vercel deployment failed**

- Does the repository contain `index.html` at the root level?
- Are all file names lowercase? Vercel runs on Linux, which is case-sensitive.
- Was the code pushed to GitHub before trying to deploy?

---

**Site works locally but not on Vercel**

Almost always caused by file name capitalisation.

Fix: rename all files to lowercase — `index.html`, `style.css`.

### Guidelines

- Do not spend more than 3 minutes on a single student's problem in front of the class.
- If 3+ students share the same issue, address the whole class.
- Ask students to read the error before asking for help: "What does it say?"
- Guide, do not write code for students. Ask: "What should this element look like?"
- Never shame a student for a basic question.
- The second instructor walks proactively — does not wait for raised hands.
